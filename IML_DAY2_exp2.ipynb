```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
x=np.array([1,1,2])
y=np.array([1,2,2])

```

```python
cofficint=np.corrcoef(x,y)
print(cofficint)
```

```python
coffecient=np.polyfit(x,y,1)
polynomial=np.poly1d(coffecient)
print(polynomial)
```

```python
error=x-y
abs_error=np.abs(error)
mae=np.mean(abs_error)
print(mae)
```
